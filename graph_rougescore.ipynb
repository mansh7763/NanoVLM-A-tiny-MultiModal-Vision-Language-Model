```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from rouge_score import rouge_scorer
```

```python
# NanoVLM-mini shortdesc

# Sample DataFrame with 25 rows (Modify as needed)
df = pd.read_csv('C:\\Users\\mansh\\OneDrive\\Desktop\\NanoVLM inferencing\\Rougescore ShortDesc\\Rougescore_shortdesc_5M.csv', encoding='ISO-8859-1')


# Initialize ROUGE scorer
scorer = rouge_scorer.RougeScorer(["rouge1"], use_stemmer=True)

# Compute ROUGE-1 scores for each row
df["rouge1_score"] = df.apply(lambda row: scorer.score(row["actual story"], row["Model Generated Story"])["rouge1"].fmeasure, axis=1)

# Plot histogram of ROUGE scores
plt.hist(df["rouge1_score"], bins=5, color="blue", alpha=0.7)
plt.xlabel("ROUGE-1 Score")
plt.ylabel("Frequency")
# plt.title("Distribution of ROUGE Scores for 25 Text Pairs")
plt.show()
```
